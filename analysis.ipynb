```python
# --- Configuration ---------------------------------------------------
USERNAME = "mnihei4452"
REPO = "MBA5278-ICP-AY2526ST2"

import os
import subprocess
import urllib.request

repo_url = f"https://github.com/{USERNAME}/{REPO}"

# Verify the fork exists and is public before attempting to clone.
try:
    urllib.request.urlopen(repo_url)
    print(f"Repository found: {repo_url}")
except Exception:
    raise RuntimeError(
        f"Cannot reach {repo_url}\n"
        "Check that: (1) USERNAME is spelled correctly, and "
        "(2) you have forked the repository on GitHub."
    )

# Clone the fork into this Colab session. subprocess is used instead of the
# "!" shell escape so that every cell remains valid Python. Colab has no
# interactive prompt, so GIT_TERMINAL_PROMPT is disabled to make a failed
# clone fail immediately rather than wait for credentials.
if not os.path.isdir(REPO):
    subprocess.run(
        ["git", "clone", f"{repo_url}.git"],
        check=True,
        env={**os.environ, "GIT_TERMINAL_PROMPT": "0"},
    )
    print(f"Cloned {REPO}")
else:
    print(f"'{REPO}' already exists in this session - skipping clone.")

# --- Load the dataset -------------------------------------------------
import pandas as pd

df = pd.read_csv(f"{REPO}/sales_recruitment_synthetic_200.csv")

print("pandas version:", pd.__version__)
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```
